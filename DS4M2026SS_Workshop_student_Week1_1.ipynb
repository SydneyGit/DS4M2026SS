Creation of variables:
- When you write a line code like `a = 10`:

In Python, variable names:

- £ can only contain letters, digits, and underscore _ (typically used to separate words in long variable names)

- cannot start with a digit. For example, 'weight' is a valid variable name, whereas 2weight is not.

- £ are case sensitive. For example, 'age', 'Age' and 'AGE' are three different variables.

- £ Variable names that start with underscores like '__alistairs_real_age' have a special meaning so we should not do that until we understand the convention.

- £ Variable names start with upper letters have special purposes too.

- £ a common variable name should look like 'part_weight', 'purchase_time_2024'

### 4.4 Type conversion

There are two types of Type Conversion in Python:
 - Implicit Type Conversion
 - Explicit Type Conversion

#### 4.4.1 Implicit Type Conversion
Python automatically converts one data type to another without any user involvement

When performing arithmetic operations between different numeric types, Python promotes the result to the more general or precise type to avoid losing information. This behaviour is called type promotion

Non-boolean values are implicitly converted to boolean in conditional statements.

#### 4.4.2 Explicit Type Conversion
The data type is manually changed by the user

To convert the integer to float, use the `float()` function in Python

To convert numbers to strings, use the `str()` function in Python

## 5. Trouble shooting

### 5.1 Common errors

Errors mean issues in the code that prevent the program from running correctly.

There are some common types of errors.

Examples are:
- misusing logic/bitwise operators: &, | ==, !=
- missing quotes, parentheses and brackets
- calling undefined functions/variables
- using the wrong indentation
- missing, misspelling, or misusing Python keywords
- ......

### 5.2 Exceptions

Exceptions are a subset of errors that occur at runtime when the program encounters an unexpected situation during execution.

Exception-based error handling mechanism has benefits of:
1. avoid program crashes
2. provide the program user meaningful feedbacks
2. `finally` block offers opportunities for running commands that should be run anyway, such as closing up files

### 5.3 Bugs

The left type of error is the most challenging to debug: logical errors (bugs). They occur when the programs run but produce wrong results. This means there is something wrong with the logic contained in the code, and you will have to review your code in order to fix it.

Here are some examples of mistakes which lead to logical errors:

   - using the wrong variable name
   - indenting a block to the wrong level
   - using integer division instead of floating-point division
   - getting operator precedence wrong
   - making a mistake in a boolean expression
   - off-by-one, and other numerical errors
